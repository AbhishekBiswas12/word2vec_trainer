```python
from word2vec_trainer.trainer import Trainer
from word2vec_trainer.preprocessing import Preprocessing
```

```python
processor = Preprocessing(
  'example_data.txt'
)
```

```python
processor.run()
```

```python
trainer = Trainer(
    vocab_size = processor.vocabulary_size,
    embedding_dim=2,
    path_to_train='train_w2v.csv',
    path_to_val='valid_w2v.csv',
    path_to_neg_dist = 'neg_distribution_test.npy',
    epochs=1000,
    lr=0.01
)
```

```python
trainer.run()
```

```python
import matplotlib.pyplot as plt

train_num = open('train_losses.txt', 'r').readlines()
val_num = open('val_losses.txt', 'r').readlines()

train_losses = [float(x.split('\t')[0]) for x in train_num]
train_pos = [float(x.split('\t')[1]) for x in train_num]
train_neg = [float(x.split('\t')[2]) for x in train_num]
val_losses = [float(x.split('\t')[0]) for x in val_num]
val_pos = [float(x.split('\t')[1]) for x in val_num]
val_neg = [float(x.split('\t')[2]) for x in val_num]
```

```python
plt.plot(train_losses)
plt.plot(val_losses)
plt.legend(['train loss', 'val loss'])
plt.show()
```

```python
plt.plot(train_pos)
plt.plot(train_neg)
plt.legend(['train pos', 'train neg'])
plt.show()
```

```python
plt.plot(val_pos)
plt.plot(val_neg)
plt.legend(['val pos', 'val neg'])
plt.show()
```

```python
import numpy as np
plt.bar(range(len(train_pos)), np.array(train_pos) - np.array(train_neg))
```

```python
plt.bar(range(len(val_pos)), np.array(val_pos) - np.array(val_neg))
```
